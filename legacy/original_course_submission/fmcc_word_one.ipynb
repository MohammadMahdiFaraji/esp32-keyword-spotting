```python
import librosa
import numpy as np
file_name=('yes_11.wav')
# بارگذاری فایل صوتی
y, sr = librosa.load(file_name, sr=16000)

# استخراج MFCC
mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)

# فلت کردن ماتریس MFCC به یک آرایه تک‌بعدی
mfcc_flat = mfcc.flatten()

# بررسی تعداد ویژگی‌ها
if len(mfcc_flat) < 650:
    mfcc_flat = np.pad(mfcc_flat, (0, 650 - len(mfcc_flat)), mode='constant')
elif len(mfcc_flat) > 650:
    mfcc_flat = mfcc_flat[:650]

print(mfcc_flat.tolist())
```
